# One hot encoding 

```python
import pandas as pd
import numpy as np
```

```python
path = "https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day27-one-hot-encoding/cars.csv"

df = pd.read_csv(path)
df.head()
```

```python
df["brand"].value_counts()
```

```python
df["fuel"].value_counts()
```

```python
df["owner"].value_counts()
```

## 1. OneHotRncoding Using Pandas

```python
pd.get_dummies(df, columns=["fuel", 'owner'], dtype=int)
```

## 2.k-1 oneHotEncoding

```python
pd.get_dummies(df, columns=["fuel", 'owner'], dtype=int, drop_first =True)
```

## 3. OneHotEncoding using sklearn

```python
X = df.iloc[:,0:4]
y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,
                                                   test_size=0.2, random_state=2)
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder 
ohe = OneHotEncoder(drop='first', sparse_output=False, dtype=np.int64)
X_train_new = ohe.fit_transform(X_train[["fuel", "owner"]])
X_test_new = ohe.transform(X_test[["fuel", "owner"]])
```

## 4. OneHotEncoding with top categories

```python
counts = df["brand"].value_counts()
```

```python
threshold = 100 
repl = counts[counts <= threshold].index
repl
```

```python
pd.get_dummies(df["brand"].replace(repl, "uncommon"), dtype=int).head()
```
