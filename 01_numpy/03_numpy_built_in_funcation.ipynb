# Creating Numpy Arrays and Matrix using Built-in Funcations

## 1. arange , linspace, logspace

````python
import numpy as np
````

````python
# range(start, stop, step)
list(range(10))
````

### 1.1 arange
```python 
np.arange(start, stop, step, dtype)

````python
np.arange(5,16, 4, dtype = np.float32)
````

````python
np.arange(6, 16, 2.5)
````

````python
np.arange(50, -1, -5)
````

### 1.2 np.linspace
```python 
np.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)

it gives values evenly spaced in linear space

````python
np.linspace(1,5, 21)
````

````python
np.linspace(1, 5, 10, dtype=np.int32)
````

````python
np.linspace(1, 5, 10, dtype=np.float32)
````

### 1.3 np.logspace
```python 
np.logspace(start, stop, num=50, endpoint=True, base=10.0, dtype=None)

 it gives values evenly spaced in logarithmic space

````python
np.logspace(1, 3, 5)  # pow(10, a)
````

````python
np.linspace(1, 1024, num=5, dtype=np.int32)
````

````python
np.logspace(0, 10, num=5, base=2, dtype=np.int32)
````
