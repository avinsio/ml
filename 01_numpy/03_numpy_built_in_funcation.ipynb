### Extract a diagona; or construct a diagonal array

### Diagonals in a Rectangular Matrix

Let `A` be an `n × m` matrix with `n` rows and `m` columns. With zero-based indexing, the top-left entry is at `(0, 0)`.

#### What is a diagonal?

A diagonal goes **down and to the right**. At each step, both the row index and the column index increase by 1.

- The **main diagonal** starts at `(0, 0)`. Its row and column indices are equal, and it has `min(n, m)` entries.
- **Upper diagonals** start in the top row, to the right of `(0, 0)`.
- **Lower diagonals** start in the first column, below `(0, 0)`.

#### How many diagonals are there?

- Main diagonal: `1`
- Upper diagonals: `m - 1`
- Lower diagonals: `n - 1`
- **Total: `n + m - 1`**

For a `4 × 3` matrix, there are **6 diagonals**: 1 main, 2 upper, and 3 lower.

#### Accessing diagonals in NumPy

Use `np.diag(A, k)` to get the diagonal with offset `k`. The offset is `column index - row index`.

- `k = 0` gives the main diagonal.
- `k = 1` gives the first diagonal above the main diagonal.
- `k = -1` gives the first diagonal below the main diagonal.

For a `4 × 3` matrix, the valid offsets are `-3, -2, -1, 0, 1, 2`.

````python
x = np.arange(9).reshape((3,3))
x
````

````python
np.diag(x)
````

````python
np.diag(x, k=1)
````

````python
np.diag(x, k = -1)
````

````python
np.diag(x, k=-2)
````
